```python
import sciencebasepy
import os
from pathlib import Path

# path where data will be downloaded
dir = Path("/beegfs/CMIP6/arctic-cmip6/cmip6_fwi/cmip6_fwi/data_release")
# login routine follows instructions here: https://code.usgs.gov/sas/sdm/sciencebasepy/-/blob/master/Authentication%20Options.ipynb
sb = sciencebasepy.SbSession()
sb.get_token()
```

```python
# the data release item on ScienceBase (https://www.sciencebase.gov/catalog/item/67ead89cd34ed02007f8357f)
item = "67ead89cd34ed02007f8357f"
item_json = sb.get_item(item)
print(item_json["title"])
print("------------------------------------------------------")
print("/n")
# get file info from item json
file_info = sb.get_item_file_info(item_json)
print(file_info)
```

```python
# filter CFFDRS files only (all in SB cloud)
indices = [6, 7, 8, 9, 10]

filenames = [file_info[i]["name"] for i in indices]
filenames
```

```python
# get download links
download_links = sb.generate_S3_download_links(item, filenames)
download_links
```

```python
# download SB Cloud files
download = sb.download_cloud_files(filenames, download_links, dir)
```
